```python
#Ejecutar función que muestra grafica de barras
def graficoBarras(label_x,label_y, titulo,atributo):
  x_values=df[atributo].unique()
  y_values=df[atributo].value_counts().tolist()
  plt.bar(x_values,y_values)
  plt.title(titulo)
  plt.xlabel(label_x)
  plt.ylabel(label_y)
  plt.show()

graficoBarras("Genero", "participación", "Participación por genero", "desc_generos")
```

```python
#Ejecutar y Mostrar histograma de la columna 'nota'
df["nota"].hist()
plt.show()
```

```python
# Muestra Grafico de proporción
tipos=df["desc_generos"].unique()
total=df["desc_generos"].value_counts().tolist()
explode=[0.4 if total[0]==max(total) else 0,0.4 if total[1]==max(total) else 0]
plt.pie(total, labels=tipos,autopct='%1.1f%%',shadow=True, explode=explode, startangle=90)
plt.title("Distribución por genero")
plt.show()
```

```python
#Muestra Grafico de densidad
x_values=df.asistencia.unique()
y_values=df.asistencia.value_counts().tolist()
plt.scatter(x_values,y_values,marker='o')
plt.title("Desidad de Asistencia")
plt.xlabel("Asistencia")
plt.ylabel("Frecuencia")
plt.show()
```
