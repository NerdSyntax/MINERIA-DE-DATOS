```python
#media de una variable numerica con respecto a una categorica
#atributo1 categorica
#atributo2 numerica
#fondo clasico=0 y dark=1
def grafica_media(df, atributo1,atributo2,title,xlabel,ylabel, fondo=0):
  if fondo==0:
    plt.style.use('fast')
  else:
    plt.style.use('dark_background')#Fondo oscuro
  x_values=df[atributo1].unique()
  y_values=df.groupby(atributo1)[atributo2].mean().tolist()
  plt.figure(figsize=(12,8))
  plt.bar(x_values,y_values)
  plt.title(title)
  plt.xlabel(xlabel)
  plt.ylabel(ylabel)
  plt.show()
```

```python
grafica_media(df,"Manufacturer","Vitamins & Minerals","Promedio de Vitamins & Minerals por Marca","Marca","Promedio de Vitaminas y Minerales", 1)
```

```python
#definición de la funcion para limpiar los nulos en atributos numericos
#La variable lista contiene los atributos a eliminarle los nulos
def delenum(lista):
  for item in lista:
    media=df[item].mean()
    df[item]=df[item].fillna(media)#fillna(bfill) ó fillna(ffill)
#definición de funcion para limpiar los nulos en atributos object
def deleobject(lista):
  for item in lista:
    moda=df[item].mode()[0]
    df[item]=df[item].fillna(moda)
```

```python
def imputarcero(lista):
  for item in lista:
    df[item]=df[item].fillna(0)
```

```python
#Ejecución de la función
delenum(['Calories','Sodium'])
deleobject(['Cereal','Manufacturer'])
```

```python
#Función que muestra grafico de barra entre una variables categorica y una numerica
#Agrupa los totales de numérica sobre la categorica
#Atributo1 categorica y Atributo2 numérica
def graficoBarras(label_x,label_y, titulo,atributo1,atributo2,fondo=0):
  plt.figure(figsize=(12,8))
  if fondo==0:
    plt.style.use('fast')
  else:
    plt.style.use('dark_background')#Fondo oscuro
  x_values=df[atributo1].unique()
  y_values=df.groupby(atributo1)[atributo2].sum().tolist()
  plt.bar(x_values,y_values)
  plt.title(titulo)
  plt.xlabel(label_x)
  plt.ylabel(label_y)
  plt.show()
```

```python
# función que grafica el top de los 10 por encima de la media
# Cuenta la variable categorica con respecto a la media de la númérica
#Atributo2 categorica y Atributo1 numérica
def grafica_pie(df,atributo1, atributo2, title, fondo=0):
  media=df[atributo1].mean()
  if fondo==0:
    plt.style.use('classic')
  else:
    plt.style.use('dark_background')#Fondo oscuro

  filtro_data = df[df[atributo1] >= media]
  top_10_ = filtro_data[atributo2].value_counts()[:10]
  plt.figure(figsize = (8,8))
  plt.title(title)
  plt.pie(top_10_ ,autopct = '%.2f',labels = top_10_.index)
  plt.show()
```

```python
grafica_pie(df,'Calories','Manufacturer',"Cereales con calorias mayor a la media", 0)
```

## **Detección de anomalias o outlier**

```python
# Usar la función vectorizada
def find_anomalies_vectorized(data):
    first_col = data.iloc[:, 0]
    data_mean = first_col.mean()
    data_std = first_col.std()
    anomaly_cut_off = data_std * 2
    lower_limit = data_mean - anomaly_cut_off
    upper_limit = data_mean + anomaly_cut_off
    print("Límite inferior:", lower_limit)
    print("Límite superior:", upper_limit)
    mask = (first_col < lower_limit) | (first_col > upper_limit)
    anomalies = first_col.index[mask].tolist()
    return anomalies

#Se selecciona la variable 'adults'  debido a que se observa en las estádistica descriptiva que el valos máximo esta muy
#alejado de la media

# Ejecutar la función
anomalies = find_anomalies_vectorized(df[['Protein']])
print("Índices de anomalías:", anomalies)

# Ver el contenido de los índices que se van a borrar
print("\nContenido de las filas que se eliminarán:")
print(df.loc[anomalies])

# Eliminar filas con anomalías
data_clean = df.drop(index=anomalies)
print("\nDataFrame limpio:")
print(data_clean)
df=data_clean
```
